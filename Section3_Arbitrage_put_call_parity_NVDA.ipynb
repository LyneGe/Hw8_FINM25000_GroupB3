## 3. Building an Options Position

### 3.1

For each of the following trades, calculate the cost of putting it on.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data_file = "option_data_bb_NVDA.xlsx"

spot = pd.read_excel(data_file, sheet_name="spot", index_col=0)
chain = pd.read_excel(data_file, sheet_name="2026-09-18")

S0 = float(spot.loc["price"].iloc[0])
contract_size = 100

def option_price(option_type, strike):
    row = chain[
        (chain["option type"] == option_type) &
        (chain["strike price"] == strike)
    ]
    return float(row["price"].iloc[0])

call_210 = option_price("Call", 210)
put_210 = option_price("Put", 210)
call_245 = option_price("Call", 245)
put_195 = option_price("Put", 195)

cost_A = contract_size * (call_210 + put_210)
cost_B = contract_size * (put_195 - call_245)
cost_C = contract_size * S0 + cost_B

cost_table = pd.DataFrame(
    {"Cost ($)": [cost_A, cost_B, cost_C]},
    index=["A", "B", "C"]
)

cost_table
```

The costs are:

| Package | Cost |
|---|---:|
| A | $3,856 |
| B | $415 |
| C | $21,279 |

### 3.2

In a sentence, describe what exposure the position has. (How would it profit or lose?)

**A.** This is a long straddle. It benefits from a large move in NVDA in either direction, while a price close to $210 results in a loss.

**B.** This is a bearish position. It profits when NVDA falls below $195, has no option payoff between $195 and $245, and loses value as NVDA rises above $245.

**C.** This is a collar. The put limits the downside below $195, while the short call caps the upside above $245.

### 3.3

Plot the payoff of each of the three strategies.

```python
stock_prices = np.linspace(100, 320, 500)

payoff_A = contract_size * (
    np.maximum(stock_prices - 210, 0) +
    np.maximum(210 - stock_prices, 0)
)

payoff_B = contract_size * (
    np.maximum(195 - stock_prices, 0) -
    np.maximum(stock_prices - 245, 0)
)

payoff_C = contract_size * stock_prices + payoff_B

plt.figure(figsize=(8, 4))
plt.plot(stock_prices, payoff_A)
plt.axvline(210, linestyle="--", linewidth=1)
plt.title("Package A Payoff")
plt.xlabel("NVDA Price at Expiration ($)")
plt.ylabel("Payoff at Expiration ($)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
plt.plot(stock_prices, payoff_B)
plt.axhline(0, linewidth=1)
plt.axvline(195, linestyle="--", linewidth=1)
plt.axvline(245, linestyle="--", linewidth=1)
plt.title("Package B Payoff")
plt.xlabel("NVDA Price at Expiration ($)")
plt.ylabel("Payoff at Expiration ($)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
plt.plot(stock_prices, payoff_C)
plt.axvline(195, linestyle="--", linewidth=1)
plt.axvline(245, linestyle="--", linewidth=1)
plt.title("Package C Payoff")
plt.xlabel("NVDA Price at Expiration ($)")
plt.ylabel("Payoff at Expiration ($)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
